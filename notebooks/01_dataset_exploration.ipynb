### Overview
This notebook demonstrates loading, parsing, and analyzing Grammatical Error Correction (GEC) datasets in official M2 format, exploring the distribution of grammatical error types according to the ERRANT taxonomy.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Sample M2 Records
sample_m2 = """
S The box of old vintage vinyl records were dropped by the movers .
A 7 8|||R:VERB:SVA|||was|||REQUIRED|||-NONE-|||0

S She will definately recieve the package untill Friday .
A 2 3|||R:SPELL|||definitely|||REQUIRED|||-NONE-|||0
A 3 4|||R:SPELL|||receive|||REQUIRED|||-NONE-|||0
A 6 7|||R:SPELL|||until|||REQUIRED|||-NONE-|||0

S The goverment provides many informations to the public .
A 1 2|||R:SPELL|||government|||REQUIRED|||-NONE-|||0
A 4 5|||R:NOUN:NUM|||information|||REQUIRED|||-NONE-|||0
"""
print("Sample M2 Records loaded successfully.")
```

```python
# Parse M2 Annotations
def parse_m2(m2_text):
    sentences = []
    current_sent = None
    for line in m2_text.strip().split("\n"):
        line = line.strip()
        if line.startswith("S "):
            current_sent = {"text": line[2:], "edits": []}
            sentences.append(current_sent)
        elif line.startswith("A ") and current_sent is not None:
            parts = line[2:].split("|||")
            span = parts[0].split()
            current_sent["edits"].append({
                "start": int(span[0]),
                "end": int(span[1]),
                "type": parts[1],
                "replacement": parts[2]
            })
    return sentences

parsed = parse_m2(sample_m2)
print(f"Parsed {len(parsed)} annotated sentences.")
print(json.dumps(parsed, indent=2))
```

```python
# ERRANT Category Distribution
all_types = [e["type"] for s in parsed for e in s["edits"]]
type_counts = Counter(all_types)
print("ERRANT Type Counts:", type_counts)

plt.figure(figsize=(8, 4))
plt.bar(type_counts.keys(), type_counts.values(), color='#6366F1')
plt.title("ERRANT Error Category Distribution")
plt.xlabel("Taxonomy Category")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

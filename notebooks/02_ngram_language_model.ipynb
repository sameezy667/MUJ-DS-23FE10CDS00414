### Overview
This notebook demonstrates building and training a Laplace-smoothed N-Gram language model, calculating word transition conditional probabilities, sentence perplexity, and fluency scoring before and after error correction.

```python
from orto.lm.ngram import NGramLanguageModel

# Corpus of standard English text
training_corpus = [
    "The box of old vintage vinyl records was dropped by the movers.",
    "She will definitely receive the package until Friday.",
    "The government provides much information to the public.",
    "There is no doubt that the committee will approve the budget.",
    "He does not like apples because they are sour.",
    "I have gone there three times to visit the museum.",
    "Despite the bad weather, they went outside.",
    "He is taller than his brother.",
    "One of my friends is coming today.",
    "I look forward to hearing from you soon.",
    "He explained the problem to me clearly.",
    "I agree with your opinion on this matter.",
    "He went to the store yesterday.",
    "She is married to a doctor."
]

model = NGramLanguageModel(n=2, smoothing=1.0)
model.train(training_corpus)
print(f"Vocabulary Size: {len(model.vocab)}")
print(f"Total N-grams: {len(model.ngram_counts)}")
```

```python
# Compute Perplexity of Erroneous vs Corrected Sentences
pairs = [
    ("The box were dropped .", "The box was dropped ."),
    ("She will recieve the package .", "She will receive the package ."),
    ("I have went there .", "I have gone there ."),
    ("She is married with a doctor .", "She is married to a doctor .")
]

print("Perplexity Comparison (Lower is more fluent):")
print("-" * 60)
for orig, corr in pairs:
    ppl_orig = model.perplexity(orig)
    ppl_corr = model.perplexity(corr)
    delta = ppl_orig - ppl_corr
    print(f"Original : {orig:<35} | PPL = {ppl_orig:.2f}")
    print(f"Corrected: {corr:<35} | PPL = {ppl_corr:.2f} | Fluency Gain: +{delta:.2f}")
    print("-" * 60)
```

### Overview
This notebook demonstrates the complete Orto diagnostic pipeline: extracting Universal Dependency priors, generating surgical hypotheses with multi-provider fallback, executing Symbolic Critic morphosyntactic verification, and in-memory reverse-offset patching.

```python
from orto.pipeline import OrtoEngine
from orto.core.syntax_engine import SyntaxEngine

engine = OrtoEngine()
syntax = SyntaxEngine()
print("OrtoEngine pipeline initialized successfully.")
```

```python
# Inspect Universal Dependency Priors
sentence = "The box of old vintage vinyl records were dropped by the movers."
priors = syntax.extract_priors(sentence)
print("Universal Dependency Priors:")
print("Key Dependencies:")
for dep in priors['key_dependencies']:
    print(f"  - {dep}")
print("\nSubject-Verb Agreement Pairs:")
for pair in priors['subject_verb_pairs']:
    print(f"  - Subject: {pair['subject']['text']} ({pair['subject']['number']}), Verb: {pair['verb']['text']} ({pair['verb']['number']}), Mismatch: {pair['agreement_mismatch']}")
```

```python
# Run Full Pipeline on Real-World Sentences
test_cases = [
    "The box of old vintage vinyl records were dropped by the movers.",
    "She will definately recieve the package untill Friday.",
    "The goverment provides many informations to the public.",
    "Their is no doubt that the committee will approve the budget.",
    "He don't likes apples because it is sour.",
    "I have went there three times.",
    "Despite of the bad weather, they went outside.",
    "He is more taller than his brother.",
    "One of my friend are coming today.",
    "I look forward to hear from you.",
    "He explained me the problem.",
    "I am agree with your opinion.",
    "She is married with a doctor."
]

for s in test_cases:
    resp = engine.analyze(s)
    print(f"ORIGINAL : {resp.original_text}")
    print(f"CORRECTED: {resp.corrected_text}")
    print(f"TIER     : {resp.telemetry.engine_tier if resp.telemetry else ''}")
    print(f"EDITS    :")
    for e in resp.edits:
        print(f"  • [{e.span.original_text} -> {e.replacement}] ({e.errant_type}) - {e.linguistic_rule} | Critic: {e.critic_verified}")
    print("-" * 70)
```
